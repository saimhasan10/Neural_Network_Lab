```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
```

#1. XOR gate classificationn using MLPClassifier

```python
X= np.array([[0,0],[0,1],[1,0],[1,1]])
y_xor=np.array([0,1,1,0])
print("No of output:", np.unique(y_xor))
print("Input size:", X.shape)

modelXor= MLPClassifier(hidden_layer_sizes=(100,),activation='relu',random_state=42,solver='sgd',alpha=0.001)
modelXor.fit(X,y_xor)
y_pred_xor=modelXor.predict(X);
print("Accuracy: ",accuracy_score(y_xor,y_pred_xor))
print("Confusion Matrix: ",confusion_matrix(y_xor,y_pred_xor))
print("Classification Report: ",classification_report(y_xor,y_pred_xor))
```

#2. Random Dataset using make classification (binary classification)

```python
from sklearn.datasets import make_classification
X, y=make_classification(n_samples=1000, n_features=10, n_classes=2);
X_train, X_test, y_train, y_test =train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
print("Y class: ", np.unique(y))
```

```python
mlp= MLPClassifier(hidden_layer_sizes=(100,10),activation='relu',random_state=42,solver='sgd',alpha=0.001)
mlp.fit(X_train,y_train)
y_pred=mlp.predict(X_test)
print("Accuracy: ",accuracy_score(y_test,y_pred))
print("Confusion Matrix: ", confusion_matrix(y_test,y_pred))
print("Classification Report: ", classification_report(y_test,y_pred))
```

#3. Random dataset for multiclass classifictaion

```python
from sklearn.datasets import make_classification
X, y=make_classification(n_samples=1000, n_features=10, n_classes=4,n_informative=4);
X_train, X_test, y_train, y_test =train_test_split(X, y, test_size=0.2, random_state=2)
print("Y class: ", np.unique(y))
```

```python
mlp= MLPClassifier(hidden_layer_sizes=(100,10),activation='relu',random_state=42,solver='sgd',alpha=0.001)
mlp.fit(X_train,y_train)
y_pred=mlp.predict(X_test)
print("Accuracy: ",accuracy_score(y_test,y_pred))
print("Confusion Matrix: ", confusion_matrix(y_test,y_pred))
print("Classification Report: ", classification_report(y_test,y_pred))
```

#2. Regression using mlpRegressor

```python
from sklearn.datasets import make_regression
from sklearn.neural_network import MLPRegressor
X, y=make_regression(n_samples=1000, n_features=10);
X_train, X_test, y_train, y_test =train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
mlp= MLPRegressor(hidden_layer_sizes=(100,10),activation='relu',random_state=42,solver='sgd',alpha=0.001)
mlp.fit(X_train,y_train)
y_pred=mlp.predict(X_test)
print("R-squared: ", mlp.score(X_test, y_test))
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse}")
```

#3. Hand written digit recognition

```python
from sklearn import datasets
digit=datasets.load_digits()
```

```python
X=digit.data
y=digit.target
print(X.shape)
print(np.unique(y))
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
mlp=MLPClassifier(hidden_layer_sizes=(100,10),activation='relu',random_state=42,solver='sgd',alpha=0.001)
mlp.fit(X_train,y_train)
y_pred=mlp.predict(X_test)
print("Accuracy: ",accuracy_score(y_test,y_pred))
print("Confusion Matrix: ", confusion_matrix(y_test,y_pred))
print("Classification Report: ", classification_report(y_test,y_pred))
```
