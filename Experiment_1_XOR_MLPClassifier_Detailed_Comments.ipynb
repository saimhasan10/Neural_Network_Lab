
# Experiment 1: XOR Gate Classification using MLPClassifier

## Objective

In this experiment, we implement a Multi-Layer Perceptron (MLP) classifier to solve the XOR classification problem.

The main purpose is to understand:
- Why a single layer perceptron cannot solve XOR
- How hidden layers introduce non-linearity
- How MLPClassifier performs binary classification

## Workflow

1. Import required libraries
2. Create XOR dataset
3. Perform Exploratory Data Analysis (EDA)
4. Configure MLPClassifier parameters
5. Train the model
6. Predict output
7. Evaluate performance using accuracy, confusion matrix, and classification report



## Cell Explanation: Library Import

Here we import all libraries required for dataset handling, visualization, model creation, and evaluation.


```python

# Import numpy for numerical array operations
import numpy as np

# Import pandas for creating and analyzing data tables
import pandas as pd

# Import matplotlib for plotting graphs
import matplotlib.pyplot as plt

# Import seaborn for advanced visualization
import seaborn as sns

# Import MLPClassifier for building the neural network classifier
from sklearn.neural_network import MLPClassifier

# Import evaluation metrics
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import roc_curve
from sklearn.metrics import auc
```


## Cell Explanation: Creating XOR Dataset

Here we manually create the XOR logic dataset.

The dataset contains:
- Two input features (X1, X2)
- One binary output class (0 or 1)


```python

# Create input features for XOR problem
# Each row represents one sample
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

# Create target labels
# XOR gives output 1 when inputs are different
y = np.array([
    0,
    1,
    1,
    0
])

# Convert data into dataframe for better visualization
data = pd.DataFrame(X, columns=["X1","X2"])

# Add target column
data["Output"] = y

# Display dataset
data
```


## Cell Explanation: Exploratory Dataset Analysis (EDA)

This section checks the dataset structure and visualizes the XOR pattern.

The visualization helps us understand why XOR is a non-linear problem.


```python

# Display dataset shape
print("Dataset Shape:", X.shape)

# Display number of input features
print("Number of Features:", X.shape[1])

# Display number of output classes
print("Number of Classes:", len(np.unique(y)))

# Display class distribution
print(data["Output"].value_counts())
```

```python

# Create scatter plot for XOR dataset
plt.figure(figsize=(6,5))

# Plot the data points with different colors for different classes
sns.scatterplot(
    x=X[:,0],
    y=X[:,1],
    hue=y,
    s=200
)

# Label x-axis
plt.xlabel("X1")

# Label y-axis
plt.ylabel("X2")

# Add title
plt.title("XOR Dataset Visualization")

# Display grid
plt.grid()

# Show plot
plt.show()
```


## Cell Explanation: MLP Model Settings

Here we define the neural network architecture.

Parameters:
- hidden_layer_sizes: number of neurons in hidden layers
- activation: activation function
- solver: optimization algorithm
- learning_rate_init: learning rate


```python

# Create MLP classifier model
model = MLPClassifier(

    # Create one hidden layer with 10 neurons
    hidden_layer_sizes=(10,),

    # Use ReLU activation function
    activation="relu",

    # Use stochastic gradient descent optimizer
    solver="sgd",

    # Set learning rate
    learning_rate_init=0.1,

    # Number of training iterations
    max_iter=1000,

    # Make results reproducible
    random_state=42
)

# Display model configuration
model
```


## Cell Explanation: Model Training

The model learns the relationship between input features and XOR output labels.


```python

# Train the neural network using XOR data
model.fit(X, y)

# Print training completion message
print("Model training completed")
```


## Cell Explanation: Prediction and Accuracy

The trained model predicts the class labels and accuracy measures the correctly classified samples.


```python

# Generate predictions using trained model
y_pred = model.predict(X)

# Display actual values
print("Actual:", y)

# Display predicted values
print("Predicted:", y_pred)

# Calculate accuracy
accuracy = accuracy_score(y, y_pred)

# Display accuracy
print("Accuracy:", accuracy)

# Display classification report
print(classification_report(y, y_pred))
```


## Cell Explanation: Confusion Matrix

Confusion matrix compares actual classes with predicted classes.


```python

# Generate confusion matrix
cm = confusion_matrix(y, y_pred)

# Create heatmap visualization
sns.heatmap(
    cm,
    annot=True,
    fmt="d"
)

# Add labels
plt.xlabel("Predicted")
plt.ylabel("Actual")

# Add title
plt.title("Confusion Matrix")

# Display matrix
plt.show()
```


## Cell Explanation: ROC Curve

ROC curve shows the classification performance at different thresholds.


```python

# Get prediction scores
y_score = model.predict_proba(X)[:,1]

# Calculate false positive rate and true positive rate
fpr, tpr, threshold = roc_curve(y, y_score)

# Calculate AUC value
roc_auc = auc(fpr, tpr)

# Plot ROC curve
plt.figure(figsize=(6,5))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {roc_auc:.2f}"
)

# Plot random classifier line
plt.plot([0,1],[0,1],"--")

# Add labels
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

# Add title
plt.title("ROC Curve")

# Show legend
plt.legend()

# Display graph
plt.show()
```


# Discussion

MLPClassifier successfully solves the XOR problem because hidden layers allow the network to learn non-linear relationships.

A single layer perceptron cannot solve XOR because it only creates a linear decision boundary.

# Conclusion

This experiment demonstrates how a multi-layer neural network overcomes the limitation of single-layer perceptrons by using hidden layers and non-linear activation functions.
